```python
import pandas as pd
```

#### Run this notebook first to construct the json file to fill db with it 

```python
df = pd.read_csv('./dataset/mozilla_core.csv')
df.head(5)
```

```python
ROWS = 100
df = df.iloc[:100]
df.head(5)
```

```python
import re
def extract_creator_name(text: str) -> str:
    prefix = "Created by "
    if prefix in text:
        # Get everything after 'Created by '
        after_prefix = text.split(prefix)[1]
        # Split by space and take the first two words (First and Last name)
        parts = after_prefix.split()
        return f"{parts[0]} {parts[1]}"
    return "Author not found"
```

```python
df["Description"] = df["Description"].fillna("Created by Unknown Author")
df['creator_name'] = df['Description'].apply(extract_creator_name)
df.head(30)
```

```python
df = df.drop(['Duplicated_issue', 'Issue_id', 'Resolved_time'], axis=1)
```

```python
df['Component'].unique()
```

```python
df['Component'].isna().sum()
```

```python

df['Component'] = df['Component'].replace('Plug-ins', 'Plugins')
```

```python
import json

# Convert timestamp-like column to datetime if needed
if 'created_time' in df.columns:
    df['Created_time'] = pd.to_datetime(df['Created_time'], errors='coerce')

json_list = []
for _, row in df.iterrows():
    item = {
        'author': row.get('creator_name', ''),
        'created_at': row.get('Created_time').isoformat() if pd.notna(row.get('created_time')) else None,
        'description': row.get('Title', ''),
        'tags': [
            row.get('Version', ''),
            row.get('Priority', ''),
            row.get('Resolution', ''),
        ],
        'module': row.get('Component', ''),
    }
    json_list.append(item)

# Save to JSON file
with open('mozilla_issues.json', 'w', encoding='utf-8') as f:
    json.dump(json_list, f, ensure_ascii=False, indent=2)

len(json_list)
```
