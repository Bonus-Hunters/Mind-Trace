```python
def count_tokens(tokenizer, text):
    return len(
        tokenizer(text, return_tensors="pt", truncation=False)["input_ids"][0]
    )

def summarize_text(summarizer, text, max_len, min_len):
    return summarizer(
        text,
        max_length=max_len,
        min_length=min_len,
        do_sample=False
    )[0]["summary_text"].strip()

def finalize_chunk(
    chunk_index,
    start_time,
    end_time,
    speakers,
    raw_text,
    summarizer=None,
    embedding_model=None,
    embedding_tokenizer=None,
    summary_max_length=150,
    summary_min_length=50
):

    summary_text = None
    embedding_vector = None

    # ---- Summarize ----
    if summarizer:
        summary_text = summarize_text(
            summarizer,
            raw_text,
            summary_max_length,
            summary_min_length
        )

    # ---- Embed ----
    if embedding_model and embedding_tokenizer:
        inputs = embedding_tokenizer(
            raw_text,
            return_tensors="pt",
            truncation=True,
            padding=True
        )
        with torch.no_grad():
            outputs = embedding_model(**inputs)
        # Mean pooling over token embeddings
        embedding_vector = outputs.last_hidden_state.mean(dim=1)[0].numpy()

    return {
        "chunk_index": chunk_index,
        "start_time_sec": start_time,
        "end_time_sec": end_time,
        "speaker_ids": list(speakers),
        "raw_text": raw_text.strip(),
        "summary_text": summary_text,
        "embedding": embedding_vector
    }
```

```python
def chunk_meeting_json(
    meeting_json,
    tokenizer,
    summarizer=None,
    embedding_model=None,
    embedding_tokenizer=None,
    max_tokens=500,
    overlap_tokens=100,
    summary_max_length=150,
    summary_min_length=50
):
    chunks = []

    current_text = ""
    current_start = None
    current_end = None
    current_speakers = set()
    chunk_index = 0

    for turn in meeting_json["dialogue"]:
        speaker = turn["speaker"]
        text = turn["text"].strip()
        start_sec = int(turn["start"])
        end_sec = int(turn["end"])

        if current_start is None:
            current_start = start_sec

        # Add turn
        current_text += f"{speaker}: {text} "
        current_end = end_sec
        current_speakers.add(speaker)

        # Token threshold
        if count_tokens(tokenizer, current_text) >= max_tokens:
            chunks.append(
                finalize_chunk(
                    chunk_index,
                    current_start,
                    current_end,
                    current_speakers,
                    current_text,
                    summarizer,
                    embedding_model,
                    embedding_tokenizer,
                    summary_max_length,
                    summary_min_length
                )
            )

            chunk_index += 1

            # Overlap
            tokens = tokenizer(
                current_text,
                return_tensors="pt",
                truncation=False
            )["input_ids"][0]

            overlap_tokens_list = tokens[-overlap_tokens:]
            overlap_text = tokenizer.decode(overlap_tokens_list)

            current_text = overlap_text + " "
            current_start = start_sec
            current_speakers = {speaker}

    # Final chunk
    if current_text.strip():
      chunks.append(
            finalize_chunk(
                chunk_index,
                current_start,
                current_end,
                current_speakers,
                current_text,
                summarizer,
                embedding_model,
                embedding_tokenizer,
                summary_max_length,
                summary_min_length
            )
        )

    return chunks
```
